#### Setup and Task 1: Tokenization

```python
from collections import Counter
from itertools import permutations
import math

train = [
    "the cat sat on the mat",
    "the cat ate the fish",
    "the dog sat on the rug",
    "the dog ate the bone"
]

test = [
    "the cat sat on the rug",
    "the fish sat on the mat"
]

# Task 1: Tokenization
def tokenize(sentence):
    words = sentence.lower().split()
    return ["<s>"] + words + ["</s>"]

train_tokens = [tokenize(s) for s in train]
test_tokens = [tokenize(s) for s in test]

print("Training tokens:")
for s in train_tokens:
    print(s)

print("\nTest tokens:")
for s in test_tokens:
    print(s)

with_markers = sum(len(s) for s in train_tokens)
without_markers = sum(len(s) - 2 for s in train_tokens)

print("\nTokens with markers:", with_markers)
print("Tokens without markers:", without_markers)


def build_vocab(sentences):
    words = set()
    for sentence in sentences:
        words.update(sentence.split())
    words.discard("<s>")
    words.add("</s>")
    return words

V_words = build_vocab(train)
V = len(V_words)

print("\nVocabulary V:", V)
print("Sorted vocabulary:", sorted(V_words))
```

#### Task 2: Generating N-grams

```python
def get_ngrams(tokens, n):
    return [tuple(tokens[i:i+n]) for i in range(len(tokens) - n + 1)]

first = train_tokens[0]

print("Unigrams:", get_ngrams(first, 1))
print("Bigrams:", get_ngrams(first, 2))
print("Trigrams:", get_ngrams(first, 3))

# Verify the number of n-grams
for sentence in train_tokens + test_tokens:
    m = len(sentence)
    for n in [1, 2, 3]:
        assert len(get_ngrams(sentence, n)) == max(0, m - n + 1)

print("All n-gram length checks passed.")

# n larger than sentence length
print("n = 10:", get_ngrams(first, 10))
```

#### Task 3: Counting N-grams

```python
unigram_counts = Counter()
bigram_counts = Counter()
trigram_counts = Counter()

for sentence in train_tokens:
    unigram_counts.update(sentence[1:])  # Exclude <s>
    bigram_counts.update(get_ngrams(sentence, 2))
    trigram_counts.update(get_ngrams(sentence, 3))

def print_counts(counts):
    for gram, count in sorted(counts.items(),
                              key=lambda x: (-x[1], x[0])):
        print(gram, ":", count)

print("Unigram counts:")
print_counts(unigram_counts)

print("\nBigram counts:")
print_counts(bigram_counts)

print("\nTrigram counts:")
print_counts(trigram_counts)

print("\nTop 5 bigrams:", bigram_counts.most_common(5))

print("Distinct unigrams:", len(unigram_counts))
print("Distinct bigrams:", len(bigram_counts))
print("Distinct trigrams:", len(trigram_counts))

def count_of(ngram):
    if isinstance(ngram, tuple):
        if len(ngram) == 1:
            return unigram_counts.get(ngram[0], 0)
        if len(ngram) == 2:
            return bigram_counts.get(ngram, 0)
        if len(ngram) == 3:
            return trigram_counts.get(ngram, 0)
    return unigram_counts.get(ngram, 0)

print("Seen bigram:", count_of(("the", "cat")))
print("Unseen bigram:", count_of(("cat", "bone")))
```

#### Task 4: Unigram Model

```python
N = sum(unigram_counts.values())

def unigram_prob(w):
    return unigram_counts.get(w, 0) / N

print("Word        Count      Probability")
for w in sorted(V_words):
    print(f"{w:<12} {unigram_counts[w]:<10} {unigram_prob(w):.4f}")

assert abs(sum(unigram_prob(w) for w in V_words) - 1) < 1e-9
print("Unigram probabilities sum to 1.")

def unigram_sentence_prob(sentence):
    tokens = tokenize(sentence)[1:]  # Exclude <s>
    return math.prod(unigram_prob(w) for w in tokens)

for s in test:
    print(s, "=", unigram_sentence_prob(s))

# Different word orders have the same unigram probability
words = test[0].split()
orders = list(permutations(words))

print("\nFirst three word orders:")
for order in orders[:3]:
    sentence = " ".join(order)
    print(sentence, "=", unigram_sentence_prob(sentence))
```

#### Task 5: Bigram Model (MLE)

```python
def context_count(prev):
    return sum(count for (v, w), count in bigram_counts.items()
               if v == prev)

def bigram_prob(prev, w):
    denominator = context_count(prev)
    if denominator == 0:
        return 0
    return bigram_counts.get((prev, w), 0) / denominator

def next_word_distribution(prev):
    denominator = context_count(prev)
    if denominator == 0:
        return {}
    return {
        w: bigram_counts.get((prev, w), 0) / denominator
        for w in sorted(V_words)
        if bigram_counts.get((prev, w), 0) > 0
    }

print("Next-word distribution for 'the':")
print(next_word_distribution("the"))

assert abs(sum(next_word_distribution("the").values()) - 1) < 1e-9

def bigram_sentence_prob(sentence):
    tokens = tokenize(sentence)
    return math.prod(
        bigram_prob(tokens[i-1], tokens[i])
        for i in range(1, len(tokens))
    )

print("\nBigram sentence probabilities:")
for s in test:
    print(s, "=", bigram_sentence_prob(s))

def predict_next(prev):
    distribution = next_word_distribution(prev)
    if not distribution:
        return None
    return sorted(distribution, key=lambda w: (-distribution[w], w))[0]

print("\nNext-word predictions:")
for word in ["the", "cat", "dog", "sat", "ate", "fish"]:
    print(word, "->", predict_next(word))
```

#### Task 6: Trigram Model

```python
def trigram_prob(w1, w2, w3):
    denominator = bigram_counts.get((w1, w2), 0)
    if denominator == 0:
        return 0
    return trigram_counts.get((w1, w2, w3), 0) / denominator

def trigram_sentence_prob(sentence):
    tokens = tokenize(sentence)

    # First word uses the bigram model; remaining tokens use trigrams
    probability = bigram_prob(tokens[0], tokens[1])

    for i in range(2, len(tokens)):
        probability *= trigram_prob(
            tokens[i-2], tokens[i-1], tokens[i]
        )

    return probability

print("Trigram sentence probabilities:")
for s in test:
    print(s, "=", trigram_sentence_prob(s))

print("\nDistinct n-grams:")
print("Unigrams:", len(unigram_counts))
print("Bigrams:", len(bigram_counts))
print("Trigrams:", len(trigram_counts))

for sentence in test_tokens:
    bigrams = get_ngrams(sentence, 2)
    trigrams = get_ngrams(sentence, 3)

    unseen_bi = [g for g in bigrams if bigram_counts.get(g, 0) == 0]
    unseen_tri = [g for g in trigrams if trigram_counts.get(g, 0) == 0]

    print("\nSentence:", " ".join(sentence[1:-1]))
    print("Unseen bigrams:", unseen_bi)
    print("Unseen trigrams:", unseen_tri)

    if len(unseen_bi) > len(unseen_tri):
        print("More unseen bigrams")
    elif len(unseen_tri) > len(unseen_bi):
        print("More unseen trigrams")
    else:
        print("Equal number of unseen bigrams and trigrams")
```

#### Task 7: Zero-Probability Problem

```python
def unseen_bigrams(sentence):
    tokens = tokenize(sentence) if isinstance(sentence, str) else sentence
    return [g for g in get_ngrams(tokens, 2)
            if bigram_counts.get(g, 0) == 0]

for s in test:
    print("\nSentence:", s)
    print("Unseen bigrams:", unseen_bigrams(s))
    print("Bigram MLE probability:", bigram_sentence_prob(s))

def safe_log(p):
    if p == 0:
        return float("-inf")
    return math.log(p)

print("\nLog probability of second test sentence:",
      safe_log(bigram_sentence_prob(test[1])))

# All possible ordered pairs of ordinary words
words_only = sorted(V_words - {"</s>"})
possible_pairs = len(words_only) ** 2

seen_word_pairs = {
    (a, b)
    for a, b in bigram_counts
    if a in words_only and b in words_only
}

unseen_pair_count = possible_pairs - len(seen_word_pairs)
percentage = unseen_pair_count / possible_pairs * 100

print("\nPossible word pairs:", possible_pairs)
print("Unseen word pairs:", unseen_pair_count)
print(f"Unseen pair percentage: {percentage:.2f}%")
```

#### Task 8: Smoothing

```python
def bigram_prob_addone(prev, w):
    return (bigram_counts.get((prev, w), 0) + 1) / (
        context_count(prev) + V
    )

print("Probability comparison:")
print("P(sat | fish):",
      f"MLE = {bigram_prob('fish', 'sat'):.4f},",
      f"Add-one = {bigram_prob_addone('fish', 'sat'):.4f}")

print("P(cat | the):",
      f"MLE = {bigram_prob('the', 'cat'):.4f},",
      f"Add-one = {bigram_prob_addone('the', 'cat'):.4f}")

# Verify smoothed probabilities sum to 1
for prev in V_words:
    total = sum(bigram_prob_addone(prev, w) for w in V_words)
    assert abs(total - 1) < 1e-9

print("All add-one probability checks passed.")

def bigram_prob_addk(prev, w, k):
    return (bigram_counts.get((prev, w), 0) + k) / (
        context_count(prev) + k * V
    )

print("\nAdd-k smoothing for P(sat | fish):")
for k in [1, 0.5, 0.1, 0.01]:
    print(f"k = {k:<5} Probability = {bigram_prob_addk('fish', 'sat', k):.4f}")
```

#### Task 9: Perplexity and Model Comparison

```python
def perplexity(sentences, prob_fn):
    log_sum = 0
    total_words = 0

    for sentence in sentences:
        tokens = tokenize(sentence)

        for i in range(1, len(tokens)):
            p = prob_fn(tokens, i)
            if p == 0:
                return float("inf")

            log_sum += math.log(p)
            total_words += 1

    return math.exp(-log_sum / total_words)


def bigram_mle_fn(tokens, i):
    return bigram_prob(tokens[i-1], tokens[i])


def bigram_addone_fn(tokens, i):
    return bigram_prob_addone(tokens[i-1], tokens[i])


def unigram_addone_fn(tokens, i):
    return (unigram_counts.get(tokens[i], 0) + 1) / (N + V)


print("Perplexity comparison:")
models = [
    ("Bigram MLE", bigram_mle_fn),
    ("Bigram Add-one", bigram_addone_fn),
    ("Unigram Add-one", unigram_addone_fn)
]

for name, fn in models:
    print(f"{name:<20} {perplexity(test, fn):.4f}")


# Compare different k values
print("\nAdd-k bigram perplexity:")
k_results = {}

for k in [1, 0.5, 0.1, 0.01]:
    def addk_fn(tokens, i, k=k):
        return bigram_prob_addk(tokens[i-1], tokens[i], k)

    score = perplexity(test, addk_fn)
    k_results[k] = score
    print(f"k = {k:<5} Perplexity = {score:.4f}")

best_test_k = min(k_results, key=k_results.get)
print("Best k on test set:", best_test_k)


# Validation: train on the first three sentences only
validation_train = train_tokens[:3]
validation_sentence = train[3]

val_unigrams = Counter()
val_bigrams = Counter()

for sentence in validation_train:
    val_unigrams.update(sentence[1:])
    val_bigrams.update(get_ngrams(sentence, 2))

def val_context_count(prev):
    return sum(c for (v, w), c in val_bigrams.items() if v == prev)

def val_bigram_addk(prev, w, k):
    return (val_bigrams.get((prev, w), 0) + k) / (
        val_context_count(prev) + k * V
    )

def validation_perplexity(sentence, k):
    tokens = tokenize(sentence)
    log_sum = 0
    count = 0

    for i in range(1, len(tokens)):
        p = val_bigram_addk(tokens[i-1], tokens[i], k)
        if p == 0:
            return float("inf")
        log_sum += math.log(p)
        count += 1

    return math.exp(-log_sum / count)

validation_results = {
    k: validation_perplexity(validation_sentence, k)
    for k in [1, 0.5, 0.1, 0.01]
}

print("\nValidation perplexity:")
for k, score in validation_results.items():
    print(f"k = {k:<5} Perplexity = {score:.4f}")

best_validation_k = min(validation_results, key=validation_results.get)
print("Best k on validation set:", best_validation_k)


# Evaluate that selected k on the test sentences,
# using only the first three training sentences.
def validation_model_test_perplexity(sentences, k):
    log_sum = 0
    count = 0

    for sentence in sentences:
        tokens = tokenize(sentence)

        for i in range(1, len(tokens)):
            p = val_bigram_addk(tokens[i-1], tokens[i], k)
            if p == 0:
                return float("inf")
            log_sum += math.log(p)
            count += 1

    return math.exp(-log_sum / count)

validation_selected_test_score = validation_model_test_perplexity(
    test, best_validation_k
)

print("\nFinal comparison:")
print("Best k selected on validation:", best_validation_k)
print("Test perplexity using validation-selected k:",
      f"{validation_selected_test_score:.4f}")
print("Best k selected directly on test:", best_test_k)
print("Best test-set perplexity using all four training sentences:",
      f"{k_results[best_test_k]:.4f}")
```
