## Activity 1

```python
import string

paragraph = "Python is powerful, simple, and useful! Data science starts with clean data."

def clean_words(text):
    words = text.lower().split()
    return [word.strip(string.punctuation) for word in words if word.strip(string.punctuation)]

cleaned_words = clean_words(paragraph)
cleaned_words
```

## Activity 2

```python
from collections import Counter

article = """
Data science uses data to understand patterns and make decisions. Data can be collected,
cleaned, analyzed, and visualized. Good data leads to better analysis, while poor data can
lead to poor decisions. Data science combines statistics, programming, and domain knowledge.
"""

words = clean_words(article)
word_counts = Counter(words)

print(f"{'Word':<15}{'Count':>5}")
print('-' * 20)
for word, count in word_counts.most_common(10):
    print(f"{word:<15}{count:>5}")
```

## Activity 3

```python
import re
import pandas as pd

paragraph = ("Data science is interesting. It combines statistics and programming. "
             "Clean data helps us build better models. Good visualization makes results easier to understand.")

sentences = [s.strip() for s in re.split(r'(?<=[.!?])\s+', paragraph) if s.strip()]
df_sentences = pd.DataFrame({'sentence': sentences})
df_sentences['word_count'] = df_sentences['sentence'].apply(lambda x: len(x.split()))
df_sentences['char_count'] = df_sentences['sentence'].apply(len)

df_sentences
```

## Activity 4

```python
import numpy as np

pair_1_a = np.array([3, 4, 0])
pair_1_b = np.array([6, 8, 0])
pair_2_a = np.array([1, 0, 1])
pair_2_b = np.array([0, 1, 1])

def cosine_similarity(a, b):
    dot_product = np.dot(a, b)
    magnitude_a = np.sqrt(np.sum(a ** 2))
    magnitude_b = np.sqrt(np.sum(b ** 2))
    return dot_product / (magnitude_a * magnitude_b)

similarity_1 = cosine_similarity(pair_1_a, pair_1_b)
similarity_2 = cosine_similarity(pair_2_a, pair_2_b)

print('Pair 1 cosine similarity:', similarity_1)
print('Pair 2 cosine similarity:', similarity_2)
print('Pair 1 is more similar because its cosine similarity is higher.')
```

```python
import json
from collections import Counter
import pandas as pd

with open('reviews.json', 'r', encoding='utf-8') as file:
    reviews = json.load(file)

df_reviews = pd.DataFrame(reviews)

def top_word(text):
    words = clean_words(text)
    return Counter(words).most_common(1)[0][0] if words else ''

df_reviews['top_word'] = df_reviews['review'].apply(top_word)
df_reviews['word_count'] = df_reviews['review'].apply(lambda x: len(clean_words(x)))
df_reviews = df_reviews.sort_values(by='word_count', ascending=False).reset_index(drop=True)

df_reviews
```
