## Task 2

```python
tokens = ["deep", "learning", "helps", "machines", "understand", "text"]
print(tokens)
print(f"Length: {len(tokens)}")
print(f"Element at index 2: {tokens[2]}\nLast Element: {tokens[-1]}") 

```

```python
tokens.append("efficiently")
print(tokens)

tokens.insert(3, "computers")
print(tokens)

tokens.remove("helps")
print(tokens)
```

```python
tokens_A = ["NLP", "uses", "Python"] 
tokens_B = ["for", "text", "analysis"]
full_tokens = tokens_A + tokens_B
print(full_tokens)
print(full_tokens[-3:])
```

```python
sorted_tokens = sorted(full_tokens)
length_sorted = sorted(full_tokens, key=len)

print("Alphabetically sorted:", sorted_tokens)
print("Length sorted:", length_sorted)
```

## Task 3

```python
word_freq = {"machine": 4, "learning": 7, "data": 5, "model": 3, "text": 6}
print(word_freq)
print(word_freq['learning'])
word_freq['data'] = 9
print(word_freq)
```

```python
word_freq['neutral'] = 8
word_freq['network'] = 5
print(word_freq)

word_freq.pop('model')
print(word_freq)

print("Found" if "text" in word_freq else "Not Found")
```

```python
vocab = {"nlp": 0, "python": 1, "token": 2, "vector": 3, "corpus": 4}
reverse = {}
for key, value in vocab.items():
    temp = key
    key = value
    value = temp

    reverse[key] = value
print(vocab)
print(reverse)
```

```python
for key, value in word_freq.items():
    print(f"Word: {key} -> Frequency: {value}")
    
print(f"No of Unique word stored in Dict: {len(word_freq)}")
```
