# Task 1

```python
import re

text = """
Junaid is studying Data Science at COMSATS in 2026.
He has 3 Python projects and scored 95 marks.
The project code is 1234 and the room number is 567.
"""

# a) Extract all digits
digits = re.findall(r"\d+", text)
print("a) Digits:", digits)

# b) Extract words that start with a capital letter
capital_words = re.findall(r"\b[A-Z][a-zA-Z]*\b", text)
print("b) Capitalized words:", capital_words)

# c) Extract sequences of exactly 4 digits
four_digits = re.findall(r"\b\d{4}\b", text)
print("c) Four-digit sequences:", four_digits)

# d) Extract words between 3 and 6 characters
short_words = re.findall(r"\b[A-Za-z]{3,6}\b", text)
print("d) Words with 3-6 characters:", short_words)
```

# Task 2

```python
text = """
Order ORD-10245 was placed by Junaid at junaid@example.com.
For assistance, contact support@gmail.com or call 0300-1234567.
The order was placed on 28/09/2026 and delivered on 30/09/2026.
The customer received a discount of 45% while another customer received 12.5%.
"""

# a) Email addresses
emails = re.findall(r"[\w.-]+@[\w.-]+\.\w+", text)
print("a) Emails:", emails)

# b) Phone numbers
phone_numbers = re.findall(r"\b03\d{2}-\d{7}\b", text)
print("b) Phone numbers:", phone_numbers)

# c) Dates
dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", text)
print("c) Dates:", dates)

# d) Percentages
percentages = re.findall(r"\b\d+(?:\.\d+)?%", text)
print("d) Percentages:", percentages)
```

# Task 3

```python
paragraph = """
I don't like complicated examples because they're difficult to understand.
It's better to start with simple NLP tasks and practice them regularly.
I'm learning Python, and I don't want to memorize everything without understanding it.
NLP is interesting because it helps computers work with human language.
"""
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

nltk_tokens = word_tokenize(paragraph)

print("NLTK Tokens:")
print(nltk_tokens)

print("Total NLTK tokens:", len(nltk_tokens))
```

```python
import spacy

nlp = spacy.blank("en")

doc = nlp(paragraph)

spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)

print("Total spaCy tokens:", len(spacy_tokens))
```

```python
print("NLTK token count:", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))
```

## Comparison:
#### 1. Both NLTK and spaCy split contractions such as "don't", "they're",
####    "it's", and "I'm" into separate tokens.
####
#### 2. The token counts are different because spaCy's blank English tokenizer
####    preserves newline characters (\n) as tokens, while NLTK does not.
####
#### 3. NLTK produced 56 tokens, while spaCy produced 61 tokens.
####    The difference of 5 tokens is due to the five newline characters.

# Task 4

```python
sentence = "Junaid bought 3 books for $50 and emailed the receipt to junaid@example.com."
nlp = spacy.blank("en")

doc = nlp(sentence)
currency_tokens = []

for token in doc:

    print(
        f"Token: {token.text:<20} "
        f"is_alpha: {token.is_alpha} "
        f"like_num: {token.like_num} "
        f"like_email: {token.like_email}"
    )

    if token.is_currency:
        currency_tokens.append(token.text)

print("\nCurrency tokens:", currency_tokens)
```

# Task 5

```python
nlp = spacy.load("en_core_web_sm")

paragraph = """
Dr. Ahmed visited the university yesterday. He met Mr. Khan in the office.
They discussed the new NLP project, e.g. the tokenization module.
The meeting ended at 4 PM. Everyone agreed to continue the discussion tomorrow.
"""

doc = nlp(paragraph)

sentences = list(doc.sents)

print("Total sentences:", len(sentences))

for number, sentence in enumerate(sentences, start=1):
    print(f"{number}. {sentence.text}")

```

### Observation:
#### spaCy correctly handled the abbreviations "Dr.", "Mr.", and "e.g."
#### without incorrectly treating their periods as sentence boundaries.
#### Therefore, the paragraph was correctly divided into 5 sentences.
