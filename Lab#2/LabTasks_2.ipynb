# Task 1

```python
import PyPDF2
import regex as re
import csv

file_path = "mathbook_linear_algebra_by_c_lay.pdf"
pdf = PyPDF2.PdfReader(file_path)
list_of_str = []
# for 10 pages
count = 0
for page in pdf.pages:
    if count==5:
        break
    list_of_str.append(page.extract_text())
    count+=1

print("Total No of Pages Extracted: ", len(list_of_str))
```

```python
merge_str = " ".join(list_of_str)
print(merge_str)

# Removing Punctuation And Digits
def removing_punc_num(text: str):
    # Remove punctuation and digits
    text = re.sub(r"[\p{P}\p{N}]+", " ", text)

    # Convert text to lowercase
    text = text.lower()

    # Split on whitespace
    tokens = text.split()

    return tokens

tokens = removing_punc_num(merge_str)
print("\nTotal Token Count:", len(tokens))
```

```python
stopwords = {
    "the", "a", "an", "and", "or",
    "is", "are", "was", "were",
    "of", "to", "in", "on", "for",
    "with", "as", "by", "at", "from",
    "this", "that", "these", "those",
    "it", "be", "been", "being",
    "which", "can"
}

filtered_tokens = []

for word in tokens:
    if word not in stopwords:
        filtered_tokens.append(word)

word_freq = {}

for word in filtered_tokens:
    if word in word_freq:
        word_freq[word] += 1
    else:
        word_freq[word] = 1

sorted_freq = sorted(
    word_freq.items(),
    key=lambda x: x[1],
    reverse=True
)

# Print top 10 words
print("\nTop 10 Most Frequent Words:")

for word, count in sorted_freq[:10]:
    print(f"{word}: {count}")
```

```python
keyword = input('Enter a keyword to search: ').lower()

found = False

for page_num, page_text in enumerate(list_of_str, start=1):

    sentences = re.split(r"(?<=[.!?])\s+", page_text)
    for sentence in sentences:
        if keyword in sentence.lower():

            print(f"\nPage {page_num}:")
            print(sentence.strip())

            found = True
if not found:
    print("NO Sentence Containing: ", keyword)
```

# Task 2

```python
pages_data = []
page_num = 0

for page in pdf.pages:
    # Read only 5 pages
    if page_num == 5:
        break
    raw_text = page.extract_text()
    if raw_text is None:
        raw_text = ""
    words = raw_text.strip().split()
    page_data = {
        'page_num'   : page_num+1,
        'raw_text'   : raw_text,
        'word_count' : len(words),
        'char_count' : len(raw_text.replace(" ", ""))   
    }
    pages_data.append(page_data)

    page_num+=1

print("\nFirst 2 Entries:")

for page in pages_data[:2]:
    
    print(page)
```

```python
non_zero_pages = [page for page in pages_data
                  if page['word_count'] > 0 ]

highest_page = max(non_zero_pages, key=lambda page: page['word_count'])
print("Highest Word Count Page\n")
print(f"Page Number: {highest_page['page_num']}")
print(f"Word Count: {highest_page['word_count']}")
print(f"Char Count: {highest_page['char_count']}")
print(f"Text: {highest_page['raw_text']}")

print("------------------------")
print("Lowest Word Count Page\n")

lowest_page = min(non_zero_pages, key= lambda page: page['word_count'])
print(lowest_page)

def get_word_count(non_zero_pages):
    total = 0
    for page in non_zero_pages:
        total += page['word_count']
    return total
sum_word_count = get_word_count(non_zero_pages)
avg_word_count = sum_word_count/len(non_zero_pages)

print(round(avg_word_count, 2))
```

```python
class PhraseNotFoundError(Exception):
    pass
try:
    search_phrase = input("Enter Search Phrase: ").lower()
    found = False
    for page in non_zero_pages:
        
        count = page['raw_text'].lower().count(search_phrase)
        if count > 0:
            found = True
            print(f"Page {page['page_num']}: Count: {count} ")
    if not found:
        raise PhraseNotFoundError(f"{search_phrase} Not found in any page")
except PhraseNotFoundError as e:
    print("\n Phrase Not Found Error: ", e)

```

```python
csv_file = "pdf_summary.csv"
with open (csv_file, 'w', newline="", encoding='utf-8') as file:
    writer = csv.writer(file)

    writer.writerow([
        'page_num',
        'word_count',
        'char_count'
    ])
    for page in pages_data:
        writer.writerow([
            page['page_num'],
            page['word_count'],
            page['char_count']
        ])
print("CSV file Written Successfully!", csv_file)

print("Reading Csv File: ")
with open(csv_file, 'r', newline="", encoding='utf-8') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python

from nltk.corpus import stopwords

english_stopwords = set(stopwords.words("english"))

english_stopwords

all_text = ""
for page in pdf.pages:
    text = page.extract_text()
    if text:
        all_text += text + " "

before_count = len(all_text)
all_text = re.sub(r"https?://\S+|www\.\S+", "", all_text)
all_text = re.sub(r"\d+", "", all_text)
all_text = re.sub(r"[^A-Za-z\s.]", "", all_text)
# all_text = re.sub(r"\s+", "", all_text)
# The 4th step is making it worse because its removing every whitespace
#  Char Count after Cleaning
after_count  = len(all_text)

print("Before Cleaning CHar Count: ", before_count)
print("After Cleaning CHar Count: ", after_count)
```

```python
cleaned_text = all_text.lower()

# Replace literal "\n" with a space
cleaned_text = cleaned_text.replace("\\n", " ")

# Replace actual newline characters with a space
cleaned_text = cleaned_text.replace("\n", " ")

# Replace multiple spaces with a single space
cleaned_text = re.sub(r"\s+", " ", cleaned_text)

# Split into sentences
sentences = re.split(r"[.!?]", cleaned_text)

print(f"Sentence Count before Removing Empty Strings: {len(sentences)}")

# Remove leading/trailing spaces and empty strings
sentences = [sentence.strip() for sentence in sentences if sentence.strip()]

print(f"Sentence Count After Removing Empty Strings: {len(sentences)}")

# Display first 5 sentences
sentences[:5]
```

```python
sentences_before_filter = len(sentences)
filtered_sentences = []

for sentence in sentences:
    tokens = sentence.split()

    filtered_tokens = []
    for word in tokens:
        if word not in english_stopwords:
            filtered_tokens.append(word)
    if len(filtered_tokens) >= 5:
        new_sentence = " ".join(filtered_tokens)
        filtered_sentences.append(new_sentence)
        
print("\nSentence count before 5-token filter:",sentences_before_filter)
print("Sentence count after 5-token filter:",len(filtered_sentences))


# Display 3 sample sentences
print("\n3 Sample Sentences:")

for sentence in filtered_sentences[:3]:
    print(sentence)

```

```python
import requests

users_url = "https://jsonplaceholder.typicode.com/users"
posts_url = "https://jsonplaceholder.typicode.com/posts"

# Send GET request for users
users_response = requests.get(users_url)

# Check status code
if users_response.status_code != 200:
    raise ConnectionError(
        f"Users API returned status code {users_response.status_code}"
    )

# Parse users JSON
users_data = users_response.json()

# Send GET request for posts
posts_response = requests.get(posts_url)

# Check status code
if posts_response.status_code != 200:
    raise ConnectionError(
        f"Posts API returned status code {posts_response.status_code}"
    )

# Parse posts JSON
posts_data = posts_response.json()

print("Users data type:", type(users_data))
print("Posts data type:", type(posts_data))

print("Number of users:", len(users_data))
print("Number of posts:", len(posts_data))
```

```python
joined_data = {}

for user in users_data:

    user_id = user["id"]
    user_name = user["name"]

    # Find posts written by this user
    user_posts = []

    for post in posts_data:
        if post["userId"] == user_id:
            user_posts.append(post)

    # Store user information and post information
    joined_data[user_name] = {
        "email": user["email"],
        "post_count": len(user_posts),
        "titles": [post["title"] for post in user_posts]
    }


# Print first 3 users only
print("First 3 users:\n")

for name, information in list(joined_data.items())[:3]:
    print(name)
    print("Email:", information["email"])
    print("Post Count:", information["post_count"])
    print("Titles:")

    for title in information["titles"]:
        print("-", title)

    print()
```

```python
# Find user with the most posts
user_with_most_posts = max(
    users_data,
    key=lambda user: sum(
        1 for post in posts_data
        if post["userId"] == user["id"]
    )
)

user_id = user_with_most_posts["id"]
user_name = user_with_most_posts["name"]

# Get this user's posts
most_user_posts = [
    post for post in posts_data
    if post["userId"] == user_id
]

print("User with the most posts")
print("------------------------")
print("Full Name:", user_name)
print("Email:", user_with_most_posts["email"])
print("Company:", user_with_most_posts["company"]["name"])

print("\nPost Titles:")

for number, post in enumerate(most_user_posts, start=1):
    print(f"{number}. {post['title']}")
```

```python
import requests

todos_url = "https://jsonplaceholder.typicode.com/todos"

response = requests.get(todos_url)

# Validate status code
if response.status_code != 200:
    print(f"Error: API returned status code {response.status_code}")
else:

    todos_data = response.json()

    # Validate that result is a list
    if not isinstance(todos_data, list):
        print("Error: API response is not a list.")

    # Validate that list is not empty
    elif len(todos_data) == 0:
        print("Error: No todo records were retrieved.")

    else:
        print("Total todo records retrieved:", len(todos_data))
```

```python
todo_summary = {}

for todo in todos_data:

    user_id = todo["userId"]

    # Create user entry if it doesn't exist
    if user_id not in todo_summary:
        todo_summary[user_id] = {
            "total": 0,
            "completed": 0,
            "incomplete": 0
        }

    # Increase total
    todo_summary[user_id]["total"] += 1

    # Check completion status
    if todo["completed"]:
        todo_summary[user_id]["completed"] += 1
    else:
        todo_summary[user_id]["incomplete"] += 1


# Print results
for user_id, data in todo_summary.items():

    print(
        f"User {user_id}: "
        f"Total = {data['total']}, "
        f"Completed = {data['completed']}, "
        f"Incomplete = {data['incomplete']}"
    )

    # Verify calculation
    check = data["completed"] + data["incomplete"] == data["total"]

    print("Verification:", check)
    print()
```

```python
completion_rates = {}

for user_id, data in todo_summary.items():

    rate = (data["completed"] / data["total"]) * 100

    completion_rates[user_id] = round(rate, 1)


# Highest completion rate
highest_user = max(
    completion_rates,
    key=completion_rates.get
)

# Lowest completion rate
lowest_user = min(
    completion_rates,
    key=completion_rates.get
)

# Overall average
average_rate = sum(completion_rates.values()) / len(completion_rates)


print(
    f"Highest completion rate: "
    f"User {highest_user} = {completion_rates[highest_user]}%"
)

print(
    f"Lowest completion rate: "
    f"User {lowest_user} = {completion_rates[lowest_user]}%"
)

print(
    f"Overall average completion rate: "
    f"{average_rate:.2f}%"
)
```

```python
users_url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(users_url)

if response.status_code != 200:
    raise ConnectionError(
        f"Users API returned status code {response.status_code}"
    )

users_data = response.json()


def flatten_user(user: dict) -> dict:

    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }


flattened_users = [
    flatten_user(user)
    for user in users_data
]

print("Flattened Users:")

for user in flattened_users:
    print(user)
```

```python

import math

df = pd.DataFrame(flattened_users)

# Convert latitude and longitude to float
df["lat"] = df["lat"].astype(float)
df["lng"] = df["lng"].astype(float)

# Calculate distance from origin
df["distance_from_origin"] = df.apply(
    lambda row: math.sqrt(row["lat"] ** 2 + row["lng"] ** 2),
    axis=1
)

print("Full DataFrame:")
print(df.to_string(index=False))


# Find user with greatest distance
farthest_user = df.loc[
    df["distance_from_origin"].idxmax()
]

print("\nUser with greatest distance from origin:")
print(farthest_user)
```

```python
# Northern Hemisphere
northern_users = df[df["lat"] > 0]

# Southern Hemisphere
southern_users = df[df["lat"] < 0]

print("Northern Hemisphere Users:")
for name in northern_users["name"]:
    print("-", name)

print("\nSouthern Hemisphere Users:")
for name in southern_users["name"]:
    print("-", name)


# Calculate average distance
northern_average = northern_users["distance_from_origin"].mean()
southern_average = southern_users["distance_from_origin"].mean()

print("\nAverage distance - Northern Hemisphere:",
      round(northern_average, 2))

print("Average distance - Southern Hemisphere:",
      round(southern_average, 2))

if northern_average > southern_average:
    print("Northern Hemisphere has the higher average distance.")
elif southern_average > northern_average:
    print("Southern Hemisphere has the higher average distance.")
else:
    print("Both hemispheres have the same average distance.")
```

```python
posts_url = "https://jsonplaceholder.typicode.com/posts"
comments_url = "https://jsonplaceholder.typicode.com/comments"

# Fetch posts
posts_response = requests.get(posts_url)

if posts_response.status_code != 200:
    raise ConnectionError(
        f"Posts API returned status code {posts_response.status_code}"
    )

posts_data = posts_response.json()


# Fetch comments
comments_response = requests.get(comments_url)

if comments_response.status_code != 200:
    raise ConnectionError(
        f"Comments API returned status code {comments_response.status_code}"
    )

comments_data = comments_response.json()


# Count comments for each post
comment_counts = {}

for comment in comments_data:

    post_id = comment["postId"]

    if post_id not in comment_counts:
        comment_counts[post_id] = 0

    comment_counts[post_id] += 1


# Create list of tuples
post_comment_list = []

for post in posts_data:

    post_id = post["id"]
    title = post["title"][:40]
    comment_count = comment_counts.get(post_id, 0)

    post_comment_list.append(
        (post_id, title, comment_count)
    )


# Sort by comment count descending
post_comment_list.sort(
    key=lambda x: x[2],
    reverse=True
)


# Top 5
print("Top 5 Most-Commented Posts")
print("---------------------------")

for post_id, title, count in post_comment_list[:5]:
    print(
        f"Post ID: {post_id} | "
        f"Comments: {count} | "
        f"Title: {title}"
    )


# Bottom 5
print("\nBottom 5 Least-Commented Posts")
print("------------------------------")

for post_id, title, count in post_comment_list[-5:]:
    print(
        f"Post ID: {post_id} | "
        f"Comments: {count} | "
        f"Title: {title}"
    )
```
